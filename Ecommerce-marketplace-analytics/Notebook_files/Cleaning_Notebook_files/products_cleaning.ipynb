```python
lookup = (
    df_o[['product_id', 'seller_id', 'list_price']]
    .drop_duplicates()
    .set_index(['product_id', 'seller_id'])['list_price']
)

preview = df_pr[df_pr['list_price'].isna()].copy()

preview['list_price_from_orders'] = preview.set_index(
    ['product_id', 'seller_id']
).index.map(lookup)

preview[['product_id', 'seller_id', 'list_price', 'list_price_from_orders']]
```

```python
preview[
    preview['list_price_from_orders'].notna()
][['product_id', 'seller_id', 'list_price_from_orders']]
```

```python
df_pr['list_price'] = df_pr['list_price'].fillna(
    pd.Series(
        df_pr.set_index(['product_id', 'seller_id'])
             .index.map(lookup),
        index=df_pr.index
    )
)
```

```python
fill_values = pd.Series(
    df_pr.set_index(['product_id', 'seller_id'])
         .index.map(lookup),
    index=df_pr.index
)

df_pr[df_pr['list_price'].isna()][
    ['product_id', 'seller_id', 'list_price']
].assign(new_price=fill_values[df_pr['list_price'].isna()])
```

```python
df_pr['list_price'] = df_pr['list_price'].fillna(fill_values)
```
