```python
import pandas as pd
import numpy as np
```

```python
reviews = pd.read_csv('reviews.csv')
```

```python
reviews.info()
```

```python
reviews.head()
```

```python
df_rw = reviews.copy()
```

```python
df_rw.info()
```

```python
df_rw.head()
```

```python
df_rw.duplicated().sum()
```

```python
df_rw = df_rw.drop_duplicates()
```

```python
df_rw.info()
```

```python
df_rw['review_date'] = pd.to_datetime(df_rw['review_date'], format = 'mixed')
```

```python
df_rw.info()
```

```python
df_rw['review_text'].unique()
```

```python
df_rw['review_text'] = df_rw['review_text'].astype(str).str.strip().str.title().replace({'Nan':np.nan})
```

```python
df_rw['review_text'].unique()
```

```python
df_rw['rating'].unique()
```

```python
df_rw['rating'] = df_rw['rating'].replace({'five': 5})
```

```python
df_rw['rating'] = pd.to_numeric(df_rw['rating'])
```

```python
df_rw['rating'].unique()
```

```python
df_rw.loc[~df_rw['rating'].between(1,5), 'rating'] = np.nan
```

```python
df_rw['rating'].unique()
```

```python
df_rw['rating'].dtypes
```

```python
rating_to_text = {5:'Excellent' , 4:'Good' , 3:'Average', 2:'Poor', 1:'Very Poor'}
```

```python
text_to_rating = {v: k for k, v in rating_to_text.items()}
```

```python
df_rw['rating'] = df_rw['rating'].fillna(df_rw['review_text'].map(text_to_rating))
```

```python
df_rw['review_text'] = df_rw['review_text'].fillna(df_rw['rating'].map(rating_to_text))
```

```python
df_rw['review_text'].unique()
```

```python
df_rw['rating'].unique()
```

```python
df_rw.info()
```

```python
df_rw['review_text'] = df_rw['review_text'].fillna('Average')
```

```python
df_rw['rating'] = df_rw['rating'].fillna(3)
```

```python
df_rw.info()
```

```python
df_rw['rating'].unique()
```

```python
df_rw['rating'] = df_rw['rating'].astype(int)
```

```python
df_rw.info()
```

```python
df_rw.to_csv('reviews_clean.csv', index = False)
```
