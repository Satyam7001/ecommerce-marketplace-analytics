```python
m = (df_o["discount_pct"].isna() & df_o["quantity"].between(1, 10))

df_o.loc[m, "discount_pct"] = (1- (df_o.loc[m, "sales_amount"]/ (df_o.loc[m, "quantity"] * df_o.loc[m, "list_price"] ))) * 100

df_o.loc[m, "discount_pct"] = (df_o.loc[m, "discount_pct"].round().clip(0, 30))
```

```python
bad = ~df_o['quantity'].between(1,10) # rows where quantity is not between 1 and 10
```

```python
bad.sum()
```

```python
Sales = df_o.loc[bad, 'sales_amount']
Listed_Price = df_o.loc[bad, 'list_price']
Discount = df_o.loc[bad, 'discount_pct']
```

```python
Quantity = Sales / (Listed_Price*(1- Discount/100))
```

```python
Quantity = Quantity.round()
```

```python
Quantity.unique()
```

```python
df_o.loc[bad, 'quantity'] = Quantity
```
