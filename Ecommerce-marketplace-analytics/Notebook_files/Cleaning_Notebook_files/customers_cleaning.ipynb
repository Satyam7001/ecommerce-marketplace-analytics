DEALING WITH AGE

```python
df_c['age'] = df_c['age'].replace({'unknown': np.nan})
```

```python
df_c['age'].unique()
```

```python
df_c['age'].isna().sum()
```

```python
df_c['age'] = pd.to_numeric(df_c['age'])
```

```python
df_c['age'].unique()
```

```python
df_c['age'].median()
```

```python
df_c['age'] = df_c['age'].fillna(df_c['age'].median())
```

```python
df_c['age'].unique()
```

```python
df_c['age'] = df_c['age'].replace([150, -5],df_c['age'].median())
```

```python
df_c['age'].unique()
```

```python
df_c['age'] = df_c['age'].astype(int)
```

```python
df_c['age'].unique()
```

```python
df_c.info()
```

DEALING WITH GENDER

```python
df_c['gender'].unique()
```

```python
df_c['gender'] = df_c['gender'].replace({'F':'Female', 'M':'Male'})
```

```python
df_c['gender'].unique()
```

```python
df_c['gender'].isna().sum()
```

```python
df_c['gender'] = df_c['gender'].fillna('Unknown')
```

```python
df_c.info()
```

```python
df_c['state'].unique()
```

```python
df_c['state'] = df_c['state'].astype(str).str.strip().str.title()
```

```python
df_c['state'].unique()
```

```python
df_c['city'].unique()
```

```python
df_c[['state', 'city']].drop_duplicates()
```

```python
# now we can se we have delhi with ony new delhi city. so we can fill nan with new delhi.
# but we can also see that there are some state with different citites. so we can't fill those. that's why we're just filling those null values with Unknown
```

```python
df_c['city'] = df_c['city'].fillna('Unknown')
```

```python
df_c['city'].unique()
```

```python
df_c.info()
```

```python
df_c = df_c.drop_duplicates()
```

```python
df_c.info()
```
